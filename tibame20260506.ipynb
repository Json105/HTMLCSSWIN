```python
import pygame as pg
import random

#pygame初始化
pg.init()

# --- 字型 ---
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)


# GAME UI
FONT_UI = get_font(32)
col, row = 15, 15
inter = 40
width, height = (col + 2) * inter, (row + 2) * inter
# 產生視窗
screen = pg.display.set_mode([width, height])
# 設定遊戲標題
pg.display.set_caption("踩地雷")

# 準備第一個圖層
bg = pg.Surface(screen.get_size())
# 把畫布填滿某個顏色
bg.fill([199, 167, 82])

# 2. 畫地雷
# pygame.draw.rect(畫布, 顏色, [x坐標, y坐標, 寬度, 高度], 線寬)
pg.draw.rect(bg, [255, 0, 0], [80, 80, inter, inter], 0)
# 3. 畫數字
pg.draw.rect(bg, [255, 255, 255], [120, 80, inter, inter], 0)
# render文字
t = FONT_UI.render("4", 1, [0, 0, 0])
# 背景(bg)上面疊上t get_rect(中心座標) -> 左上角座標
bg.blit(t, t.get_rect(center=[140, 100]))

# 1. 畫棋盤
# pygame.draw.line(畫布, 顏色, (x坐標1, y坐標1), (x坐標2, y坐標2), 線寬)
for i in range(row+1):
    pg.draw.line(bg,
                 [0, 0, 0],
                 [inter, inter*i+inter],
                 [width-inter, inter*i+inter],
                 1)

for i in range(col+1):
    pg.draw.line(bg,
                 [0, 0, 0],
                 [inter*i+inter, inter],
                 [inter*i+inter, height-inter],
                 1)

# 你要把圖層放到上一層
screen.blit(bg, [0, 0])
# 對畫面進行更新(才會真的秀出來)
pg.display.update()


# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測滑鼠點擊以後放掉的動作
        if event.type == pg.MOUSEBUTTONUP:
            pass
        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False

pg.quit()
```

```python
def print2hello():
    for j in range(2):
        print("hello")

for i in range(3):
    print("第", i, "次")
    print2hello()
```

```python
# 多層迴圈(先寫裡面)
# 第一次
# hello
# hello
# 第二次
# hello
# hello
for i in range(3):
    print("第", i, "次")
    for j in range(2):
        print("hello")
```

```python
# oxoxo
# xoxox
# oxoxo
# xoxox
for i in range(4):
    line = ""
    for j in range(5):
        if i % 2 == 0:
            if j % 2 == 0:
                line = line + "o"
            else:
                line = line + "x"
        else:
            if j % 2 == 0:
                line = line + "x"
            else:
                line = line + "o"
    print(line)
```

```python
for i in range(4):
    line = ""
    for j in range(5):
        # j = 0 2 4 6 8
        if (i + j) % 2 == 0:
            line = line + "o"
        # j = 1 3 5 7 9
        else:
            line = line + "x"
    print(line)
```

```python
symbols = ["o", "x", "@"]

for i in range(5):
    startpos = i % len(symbols)
    line = ""
    for j in range(7):
        pos = (startpos + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```
